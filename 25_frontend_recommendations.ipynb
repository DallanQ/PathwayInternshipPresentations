# Frontend Recommendations

These are just my opinions on tools to use to create frontend websites if you are a backend developer.

### 1. Are you ok with a simple UI?

Use Gradio (simplest) or Streamlit (more complex, for dashboards) and host your app on HuggingFace spaces.

### 2. Are you building a chatbot?

Use the Llamaindex [create-llama](https://www.npmjs.com/package/create-llama) tool to create a frontend and a backend that you can customize.

### 3. Can you live with static HTML?

Use [flask](https://flask.palletsprojects.com/) (mature, large community), or [FastHTML](https://www.fastht.ml/) (easier to learn, but brand new and no real community).

### 4. Know React, or want to learn the most-popular javascript framework?

Use [React](https://react.dev/) for the frontend and [FastAPI](https://fastapi.tiangolo.com/) for the backend. (These are the tools that the create-llama tool uses.)

### 5. Otherwise

Use [Vue](https://vuejs.org/) (mature, large community) or [Svelte and Sveltekit](https://kit.svelte.dev/) (nicer syntax but less mature and a bit more difficult to deploy with FastAPI) for the frontend and  [FastAPI](https://fastapi.tiangolo.com/) for the backend.

[Here is an example app](https://github.com/notarious2/Vue3FastAPI-To-Do) that uses Vue and FastAPI.

## Hosting options

1. Render seems to work well for simple projects.
2. AWS copilot also works well, but AWS has a steep (but probably worthwhile) learning curve.
3. I had problems hosting FastAPI on Vercel, but your mileage may vary.
